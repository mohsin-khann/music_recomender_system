```python
import pandas as pd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
```

```python
df=pd.read_csv("/content/spotify_songs.csv")
```

```python
df.head(12).columns
```

```python
df.drop(columns=['acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo',
       'duration_ms', 'language'],inplace=True)
```

```python
df
```

```python
df.drop(columns=['danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness'],inplace=True)
```

```python
data=df[['track_name','track_artist','lyrics','playlist_name']]
```

```python
# Fill missing values for relevant columns
data['track_name'] = data['track_name'].fillna('')
data['track_artist'] = data['track_artist'].fillna('')
data['lyrics'] = data['lyrics'].fillna('')
data['playlist_name'] = data['playlist_name'].fillna('')
```

```python
# Function to combine track name, artist, lyrics, and playlist name
def combine_features(row):
    return row['track_name'] + ' ' + row['track_artist'] + ' ' + row['lyrics'] + ' ' + row['playlist_name']

# Apply the function to create the 'combined_features' column
data['combined_features'] = data.apply(combine_features, axis=1)

# Initialize the TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(data['combined_features'])
```

```python
# Initialize the NearestNeighbors model
nn_model = NearestNeighbors(metric='cosine', algorithm='brute')

# Fit the model on the tfidf_matrix
nn_model.fit(tfidf_matrix)
```

```python
# Function to get recommendations
def get_music_recommendations(track_name, n=8):
    # Transform the input feature to the same vector space as the tfidf_matrix
    input_vector = tfidf_vectorizer.transform([track_name])

    # Find the nearest neighbors for the input vector
    distances, indices = nn_model.kneighbors(input_vector, n_neighbors=n + 1)

    # Get the details of the top n most similar tracks
    similar_tracks = data.iloc[indices.flatten()[1:]]  # Exclude the first one as it is the input track itself

    return similar_tracks
```

```python
recommended_tracks = get_music_recommendations('All i want ', n=5)
print(recommended_tracks)
```
